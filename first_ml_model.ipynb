First ML Model

Load Data

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
df
```

Data preparation

Data seperation as X and Y

```python
y = df['logS']
y
```

```python
X = df.drop('logS', axis=1)
X
```

Data splitting

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(X,y, test_size =0.2 , random_state = 100)
```

```python
X_train
```

```python
X_test
```

Model Building

Linear regression


```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
```

Applying the model to make a prediction


```python
y_lr_train_pred = lr.predict(X_train)
y_lr_test_pred = lr.predict(X_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test_pred
```

Evaluate model Performance

```python
from sklearn.metrics import mean_squared_error , r2_score

lr_train_mne = mean_squared_error(y_train , y_lr_train_pred)
lr_train_r2 = r2_score(y_train,y_lr_train_pred)

lr_test_mne = mean_squared_error(y_test,y_lr_test_pred)
lr_test_r2 = r2_score(y_test , y_lr_test_pred)
```

```python
print('LR MSE (Train) :' , lr_train_mne)
print('LR R2 (Train) :' , lr_train_r2)
print('LR MSE (Test) :' , lr_test_mne)
print('LR R2 (Test) :' , lr_test_r2)
```

```python
lr_results = pd.DataFrame(['LinearRegression' , lr_train_mne , lr_train_r2 , lr_test_mne , lr_test_r2]).transpose()
lr_results.columns = ['Method' , 'Training MSE' , 'Training R2' , 'Test MSE' , 'Test R2']
```

```python
lr_results
```

Random Forest

Training the model

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(max_depth=2 , random_state=100)
rf.fit(X_train, y_train)
```

Applying the model to make a prediction

```python
y_rf_train_pred = rf.predict(X_train)
y_rf_test_pred = rf.predict(X_test)
```

Evaluate model performance


```python
from sklearn.metrics import mean_squared_error , r2_score

rf_train_mne = mean_squared_error(y_train , y_rf_train_pred)
rf_train_r2 = r2_score(y_train,y_rf_train_pred)

rf_test_mne = mean_squared_error(y_test,y_rf_test_pred)
rf_test_r2 = r2_score(y_test , y_rf_test_pred)
```

```python
rf_results = pd.DataFrame(['Random Forest' , rf_train_mne , rf_train_r2 , rf_test_mne , rf_test_r2]).transpose()
rf_results.columns = ['Method' , 'Training MSE' , 'Training R2' , 'Test MSE' , 'Test R2']
rf_results
```

Model Comparison

```python
df_models = pd.concat([lr_results , rf_results], axis = 0)
df_models
```

```python
df_models.reset_index(drop=True)
```

Data Visualization of prediction results

```python
import matplotlib.pyplot as plt
import numpy as np

plt.Figure(figsize=(5,5))
plt.scatter(x=y_train , y=y_lr_train_pred , c="#0E00AE" ,alpha=0.3)

z= np.polyfit(y_train , y_lr_train_pred, 1)
p = np.poly1d(z)

plt.plot(y_train, p(y_train), '#E8766D')
plt.ylabel('Predict Logs')
plt.xlabel('Experimental LogS')
```
